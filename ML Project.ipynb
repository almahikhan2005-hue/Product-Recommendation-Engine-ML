```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

data = {
    'product_name': [
        'Hydrating Face Cream', 'Vitamin C Serum', 'Sunscreen SPF 50', 
        'Moisturizing Lotion', 'Brightening Serum', 'Matte Red Lipstick', 
        'Liquid Foundation', 'Waterproof Eyeliner', 'Nude Lip Gloss', 
        'Full Coverage Concealer', 'Volumizing Mascara'
    ],
    'description': [
        'water based hydrating cream for dry skin and face moisture',
        'vitamin c facial serum for brightening glow and skin repair',
        'water resistant sunscreen spf 50 for sun protection and skin care',
        'daily moisturizing body lotion for dry skin hydration',
        'concentrated brightening serum with vitamin c and antioxidants',
        'long lasting matte red lipstick with intense pigment',
        'full coverage liquid foundation for smooth skin tone and makeup base',
        'black waterproof eyeliner for long wear eye makeup',
        'shiny nude lip gloss for glowing lips and makeup finish',
        'liquid concealer for full coverage of dark circles and spots',
        'black volumizing mascara for thick eyelashes and eye definition'
    ]
}

df = pd.DataFrame(data)
 
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['description'])
cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)

def get_recommendations(product_name, cosine_sim=cosine_sim):
    try:

        idx = df[df['product_name'].str.lower() == product_name.lower()].index[0]
        
        sim_scores = list(enumerate(cosine_sim[idx]))
        sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
        
        sim_indices = [i[0] for i in sim_scores[1:4]]
        
        return df['product_name'].iloc[sim_indices]
    except IndexError:
        return "Product not found. Please check the spelling!"

print("--- Product List ---")
print(", ".join(df['product_name'].tolist()))
print("-" * 20)

user_choice = input("Which product are you looking at? ")
results = get_recommendations(user_choice)

print(f"\nSince you liked '{user_choice}', you might also like:")
print(results)
```
